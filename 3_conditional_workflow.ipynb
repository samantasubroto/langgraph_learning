```python
from typing import TypedDict, Literal
from langgraph.graph import StateGraph, START, END
```

```python
class State(TypedDict):
    number: int
    result: str
```

```python
def even_node(state: State):
    return {"result": f"{state['number']} is even"}
```

```python
def odd_node(state: State):
    return {"result": f"{state['number']} is odd"}
```

```python
def route(state: State) -> Literal["even_node", "odd_node"]:
    return "even_node" if state["number"] % 2 == 0 else "odd_node"
```

```python
# 4. Build the graph
graph = StateGraph(State)
graph.add_node("even_node", even_node)
graph.add_node("odd_node", odd_node)

graph.add_conditional_edges(START, route)   # the conditional step
graph.add_edge("even_node", END)
graph.add_edge("odd_node", END)

workflow = graph.compile()
```

```python
print(workflow.invoke({"number": 315143513451351345134514351345562456451351}))
```
